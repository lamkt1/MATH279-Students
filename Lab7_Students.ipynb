A covariance matrix is a square matrix that shows the covariance between every pair of variables in a dataset.


*   Positive covariance means two variables tend to increase together.
*   Negative covariance means one tends to increase while the other decreases.
*   Covariance near zero suggests little linear relationship.

Consider the variables `total sulfur dioxide` and `citric acid`, the total sulfur dioxide has a much larger numerical scale, so its variance will be much larger than the variance of citric acid. The covariance matrix can be dominated by high-variance variables. Therefore, standardization is recommended before computing a covariance matrix.

Recall from Statistics course that data standardization transforms each variable to have mean µ = 0 and standard deviation σ = 1. After standardization, all variables are on the same scale.

When data are standardized, the covariance matrix becomes closely related to the correlation matrix.




```python
# Standardize the data
X_standardized = (X - X.mean()) / X.std()

# Display the first five rows of X_standardized 👇👇👇
```

## Principal Component Analysis (PCA)

PCA starts by computing the covariance matrix and then finding its eigenvalues and eigenvectors.

The covariance matrix reveals:

*   Directions of maximum variation.
*   Redundant variables that move together.

PCA uses this information to reduce dimensionality while retaining most of the information.

```python
# Compute the covariance matrix
# Linear relationships between variables
# Resulting covariance matrix is always a square matrix
C = X_standardized.cov()
C
```
